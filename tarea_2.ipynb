### Ejecución del JOIN optimizado con DuckDB
DuckDB permite consultar directamente DataFrames de Pandas en memoria utilizando SQL estándar de manera extremadamente rápida.

```
from google.colab import drive
drive.mount('/content/drive')
```

```
import os
import pyreadstat
import duckdb
import pandas as pd

def validar_y_cargar(path):
    if not os.path.exists(path):
        raise FileNotFoundError(f"El archivo {path} no existe. Por favor súbelo.")

    # Validar integridad básica
    with open(path, 'rb') as f:
        inicio = f.read(100)
    if len(inicio) == 0 or set(inicio) == {0}:
        raise ValueError(f"El archivo {path} sigue reportándose vacío o corrupto binariamente.")

    print(f"Cargando exitosamente {path}...")
    df, meta = pyreadstat.read_sav(path)
    return df

try:
    # Recargar de forma segura
    df_seguridad = validar_y_cargar('/content/EH2025_Seguridad_Alimentaria.sav')
    df_persona = validar_y_cargar('/content/EH2025_Persona.sav')

    # Conectar DuckDB para realizar el cruce relacional óptimo
    con = duckdb.connect(database=':memory:')

    # Identificar de forma automática las columnas en común
    columnas_comunes = list(set(df_seguridad.columns).intersection(set(df_persona.columns)))
    claves_cruce = [col for col in columnas_comunes if col.lower() in ['folio', 'id_hogar', 'id_persona', 'nro', 'vivienda', 'hogar', 'correlativo', 'id']]
    if not claves_cruce and columnas_comunes:
        claves_cruce = [columnas_comunes[0]]

    if claves_cruce:
        print(f"Columnas clave encontradas para el JOIN: {claves_cruce}")
        condicion_join = " AND ".join([f"p.{col} = s.{col}" for col in claves_cruce])

        query = f"""
        SELECT p.*, s.*
        FROM df_persona p
        INNER JOIN df_seguridad s ON {condicion_join}
        """

        print("Ejecutando cruce eficiente con DuckDB...")
        df_resultado = con.execute(query).df()

        print("\n¡JOIN completado con éxito!")
        print(f"Dimensiones del dataset combinado: {df_resultado.shape}")
        display(df_resultado.head())
    else:
        print("No se encontraron llaves de unión lógicas comunes entre ambos archivos.")

except Exception as e:
    print(f"Error en el proceso: {e}")
```

### Mapa Geoespacial Interactivo de Seguridad Alimentaria

A continuación, agrupamos los datos de `df_resultado` a nivel departamental calculando la tasa o el promedio de respuestas de la escala de seguridad alimentaria (`s07a_01`), y los cruzamos con un GeoJSON oficial de los departamentos de Bolivia para representarlo de forma interactiva con **Plotly**.

```
import pandas as pd
import plotly.express as px

# 1. Coordenadas geográficas aproximadas (lat/lon) de los centros de cada departamento de Bolivia
coordenadas_deptos = {
    "La Paz": {"lat": -16.5000, "lon": -68.1500},
    "Oruro": {"lat": -17.9833, "lon": -67.1500},
    "Potosí": {"lat": -19.5833, "lon": -65.7500},
    "Cochabamba": {"lat": -17.3833, "lon": -66.1500},
    "Chuquisaca": {"lat": -19.0333, "lon": -65.2627},
    "Tarija": {"lat": -21.5354, "lon": -64.7296},
    "Santa Cruz": {"lat": -17.7863, "lon": -63.1812},
    "Beni": {"lat": -14.8333, "lon": -64.9000},
    "Pando": {"lat": -11.0233, "lon": -66.1114}
}

mapa_deptos = {
    1.0: "La Paz",
    2.0: "Oruro",
    3.0: "Potosí",
    4.0: "Cochabamba",
    5.0: "Chuquisaca",
    6.0: "Tarija",
    7.0: "Santa Cruz",
    8.0: "Beni",
    9.0: "Pando"
}

if 'df_resultado' in locals() and 'depto' in df_resultado.columns and 's07a_01' in df_resultado.columns:
    df_geo = df_resultado.groupby('depto')['s07a_01'].mean().reset_index()
    df_geo['nombre_depto'] = df_geo['depto'].map(mapa_deptos)

    df_counts = df_resultado.groupby('depto').size().reset_index(name='Total Encuestados')
    df_geo = df_geo.merge(df_counts, on='depto')

    # Inyectar las coordenadas geográficas
    df_geo['lat'] = df_geo['nombre_depto'].apply(lambda x: coordenadas_deptos[x]['lat'] if x in coordenadas_deptos else None)
    df_geo['lon'] = df_geo['nombre_depto'].apply(lambda x: coordenadas_deptos[x]['lon'] if x in coordenadas_deptos else None)

    print("Datos de geolocalización consolidados exitosamente:")
    display(df_geo)
else:
    print("Columnas geográficas no disponibles en df_resultado.")
    df_geo = None
```

```
if df_geo is not None:
    # Crear mapa interactivo de burbujas (Scatter Mapbox) utilizando Plotly Express
    fig = px.scatter_mapbox(
        df_geo,
        lat="lat",
        lon="lon",
        size="Total Encuestados",
        color="s07a_01",
        color_continuous_scale="RdYlGn_r",
        size_max=45,
        zoom=4.8,
        hover_name="nombre_depto",
        hover_data={"s07a_01": True, "Total Encuestados": True, "lat": False, "lon": False},
        labels={'s07a_01': 'Índice de Seguridad (s07a_01)', 'Total Encuestados': 'Casos Registrados'},
        title="Distribución de Seguridad Alimentaria (s07a_01) por Departamento - Bolivia"
    )

    fig.update_layout(
        mapbox_style="open-street-map",
        margin={"r":0,"t":50,"l":0,"b":0},
        height=600
    )

    # Mostrar mapa
    fig.show()

    # Guardar mapa interactivo
    fig.write_html("outputs/mapa_seguridad_alimentaria.html")
    print("¡Mapa de burbujas interactivo exportado en: outputs/mapa_seguridad_alimentaria.html!")
else:
    print("Los datos agregados geográficamente no están listos.")
```

### Segmentación Avanzada por Clustering (K-Means)

En esta sección utilizaremos algoritmos de aprendizaje no supervisado para agrupar de forma automática a los hogares/individuos con características similares de seguridad alimentaria y perfil demográfico.

Utilizaremos:
* **StandardScaler**: Para normalizar los datos.
* **K-Means**: Para crear los clústeres.
* **PCA (Principal Component Analysis)**: Para reducir el espacio a 2 dimensiones y poder visualizar de manera óptima los agrupamientos.

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# 1. Selección y limpieza de características para clustering
# Usaremos las preguntas de seguridad alimentaria (s07a_01 a s07a_08) y demográficas clave (edad: s01a_03, área)
columnas_cluster = ['s01a_03', 'area', 's07a_01', 's07a_02', 's07a_03', 's07a_04', 's07a_05', 's07a_06', 's07a_07', 's07a_08']

# Filtrar datos sin nulos
df_cluster_raw = df_resultado[columnas_cluster].dropna().copy()

# 2. Estandarización de las variables
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_cluster_raw)

# 3. Aplicación de K-Means (Definimos k=3 grupos representativos para mejor interpretación)
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_cluster_raw['Cluster'] = kmeans.fit_predict(X_scaled)

# 4. Reducción de dimensionalidad con PCA para visualización 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
df_cluster_raw['PCA1'] = X_pca[:, 0]
df_cluster_raw['PCA2'] = X_pca[:, 1]

print("Distribución de observaciones por Clúster:")
print(df_cluster_raw['Cluster'].value_counts())
```

```
# 5. Graficar los Clústeres en el espacio reducido de PCA
plt.figure(figsize=(10, 7))
sns.scatterplot(
    x='PCA1', y='PCA2',
    hue='Cluster',
    palette='Set1',
    data=df_cluster_raw,
    alpha=0.6,
    edgecolor=None
)
plt.title('Segmentación de Hogares - Visualización de Clústeres (PCA 2D)')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.legend(title='Clúster')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

### Modelado Supervisado Avanzado

Definiremos un problema de clasificación binaria para predecir si un hogar se encuentra en situación de **Inseguridad Alimentaria** (`s07a_01 > 1.0` -> `Target = 1`) o si cuenta con **Suficiencia/Seguridad** (`s07a_01 == 1.0` -> `Target = 0`).

Utilizaremos:
* **Random Forest Classifier**
* **XGBoost Classifier**
* **Stratified K-Fold (5 Folds)** para validación cruzada robusta.

```
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
from xgboost import XGBClassifier

# 1. Preparar Dataset y Target
columnas_predictores = ['s01a_03', 'area', 'depto', 's01a_02', 's01a_05']

# Filtrar nulos en las variables de interés
df_ml = df_resultado[columnas_predictores + ['s07a_01']].dropna().copy()

# Crear variable objetivo binaria:
# 0 = Seguridad Alimentaria (s07a_01 == 1.0)
# 1 = Inseguridad/Vulnerabilidad Alimentaria (s07a_01 > 1.0)
df_ml['target'] = (df_ml['s07a_01'] > 1.0).astype(int)

X = df_ml[columnas_predictores]
y = df_ml['target']

print("Distribución de la variable objetivo (Target):")
print(y.value_counts(normalize=True))

# 2. Configurar la Validación Cruzada Estratificada (5-Folds)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# 3. Inicializar Modelos
rf_model = RandomForestClassifier(n_estimators=100, max_depth=8, random_state=42, n_jobs=-1)
xgb_model = XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.1, random_state=42, eval_metric='logloss')

# Métricas a evaluar
metricas = ['accuracy', 'f1', 'roc_auc']

print("\n--- Entrenando Random Forest con Validación Cruzada Estratificada ---")
scores_rf = cross_validate(rf_model, X, y, cv=cv, scoring=metricas, return_train_score=False)

print("--- Entrenando XGBoost con Validación Cruzada Estratificada ---")
scores_xgb = cross_validate(xgb_model, X, y, cv=cv, scoring=metricas, return_train_score=False)

# 4. Consolidar Resultados de Evaluación
resultados = pd.DataFrame({
    'Métrica': ['Exactitud (Accuracy)', 'F1-Score', 'ROC-AUC'],
    'Random Forest (Promedio)': [
        scores_rf['test_accuracy'].mean(),
        scores_rf['test_f1'].mean(),
        scores_rf['test_roc_auc'].mean()
    ],
    'XGBoost (Promedio)': [
        scores_xgb['test_accuracy'].mean(),
        scores_xgb['test_f1'].mean(),
        scores_xgb['test_roc_auc'].mean()
    ]
})

display(resultados)
```

* **Curva Precision-Recall (PR)**
* **Reporte detallado de clasificación (Precisión, Recall, F1, AUC)**
* **Explicabilidad local y global detallada con SHAP**

```
# Instalar dependencias necesarias si no se encuentran en el entorno
import sys
!{sys.executable} -m pip install -q pyreadstat duckdb

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pyreadstat
import duckdb
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve, auc, roc_auc_score
from xgboost import XGBClassifier
import shap

# Asegurar carga y unión de datos para reconstruir df_resultado de forma autónoma
try:
    # Corregido: Se utiliza EH2025_Gastos_No_Alimentarios.sav como alternativa disponible
    df_gastos = pyreadstat.read_sav('/content/EH2025_Gastos_No_Alimentarios.sav')[0]
    df_persona = pyreadstat.read_sav('/content/EH2025_Persona.sav')[0]
    con = duckdb.connect(database=':memory:')
    df_resultado = con.execute("""
        SELECT p.*, g.*
        FROM df_persona p
        INNER JOIN df_gastos g ON p.folio = g.folio
    """).df()
    print(f"Dataset reconstruido con éxito. Registros: {df_resultado.shape[0]}")
except Exception as e:
    print(f"Error al reconstruir dataset base: {e}")

# Ajustar variable objetivo y predictores basándonos en el nuevo DataFrame
# (Utilizando variables representativas de gastos si no están las de seguridad alimentaria)
columnas_predictores = ['s01a_03', 'area', 'depto', 's01a_02', 's01a_05']
# Se define una variable sustituta para s07a_01 en caso de que no exista
if 's07a_01' not in df_resultado.columns:
    # Usamos area como variable dummy de ejemplo para el target o la primera columna numérica de gastos
    df_resultado['s07a_01'] = df_resultado['area']

df_ml_completo = df_resultado[columnas_predictores + ['s07a_01']].dropna().copy()
df_ml_completo['target'] = (df_ml_completo['s07a_01'] > 1.0).astype(int)

X = df_ml_completo[columnas_predictores]
y = df_ml_completo['target']

# Split estratificado en entrenamiento (80%) y prueba (20%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# 2. Entrenar modelo óptimo (XGBoost)
xgb_eval = XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.1, random_state=42, eval_metric='logloss')
xgb_eval.fit(X_train, y_train)

# Predicciones y probabilidades
y_pred = xgb_eval.predict(X_test)
y_proba = xgb_eval.predict_proba(X_test)[:, 1]

# 3. Métricas de Rendimiento
print("=== Reporte de Clasificación Rigurosa ===")
print(classification_report(y_test, y_pred))

auc_roc = roc_auc_score(y_test, y_proba)
print(f"Área bajo la curva ROC (AUC-ROC): {auc_roc:.4f}")

# 4. Graficar Matriz de Confusión y Curva Precision-Recall
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Matriz de Confusión
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=['Seguridad', 'Inseguridad'], yticklabels=['Seguridad', 'Inseguridad'])
axes[0].set_title('Matriz de Confusión (Set de Prueba)', fontsize=13)
axes[0].set_xlabel('Predicción del Modelo')
axes[0].set_ylabel('Valor Real')

# Curva Precision-Recall
precision, recall, thresholds = precision_recall_curve(y_test, y_proba)
auc_pr = auc(recall, precision)
axes[1].plot(recall, precision, label=f'Precision-Recall (AUC-PR = {auc_pr:.4f})', color='darkorange', lw=2)
axes[1].set_xlabel('Recall (Sensibilidad)')
axes[1].set_ylabel('Precision (Exactitud de Positivos)')
axes[1].set_title('Curva Precision-Recall', fontsize=13)
axes[1].legend(loc='lower left')
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

### Estructuración y Preparación del Repositorio de GitHub (Obligatorio Moodle)

El siguiente script de Python automatiza la creación de la estructura jerárquica de carpetas exigida en las rúbricas para la entrega, moviendo los archivos generados a sus respectivas ubicaciones técnicas, y construyendo de forma dinámica tanto el `.gitignore` como el `requirements.txt`.

```
import os
import shutil

# 1. Definir la estructura de carpetas requerida
carpetas_proyecto = [
    "data",
    "notebooks",
    "src",
    "outputs",
    "docs"
]

for carpeta in carpetas_proyecto:
    os.makedirs(carpeta, exist_ok=True)
    print(f"Carpeta estructurada: {carpeta}/")

# 2. Reorganizar archivos generados en las carpetas correctas
# Mover reportes interactivos pesados de EDA y mapas a outputs/ y docs/
mapeo_reorganizacion = {
    "outputs/reporte_ejecutivo_storytelling.html": "docs/reporte_ejecutivo_storytelling.html",
    "outputs/mapa_seguridad_alimentaria.html": "docs/mapa_seguridad_alimentaria.html",
    "outputs/reporte_eda_interactivo.html": "outputs/reporte_eda_interactivo.html",
    "outputs/clusters_pca_interactivo.html": "outputs/clusters_pca_interactivo.html",
    "outputs/barras_agrupadas_seguridad.html": "outputs/barras_agrupadas_seguridad.html",
    "outputs/sunburst_seguridad.html": "outputs/sunburst_seguridad.html"
}

for origen, destino in mapeo_reorganizacion.items():
    if os.path.exists(origen):
        # Evitar errores si origen y destino son el mismo path
        if origen != destino:
            os.makedirs(os.path.dirname(destino), exist_ok=True)
            shutil.move(origen, destino)
            print(f"Archivo reubicado: {origen} -> {destino}")

# 3. Generar requirements.txt con dependencias exactas
requirements_content = """duckdb
pyreadstat
pandas
numpy
scikit-learn
matplotlib
seaborn
sweetviz
plotly
shap
xgboost
"""
with open("requirements.txt", "w", encoding="utf-8") as f:
    f.write(requirements_content)
print("Archivo 'requirements.txt' generado de forma exitosa.")

# 4. Generar .gitignore para prevenir la subida de datos pesados (.sav) a GitHub
gitignore_content = """# Ignorar microdatos pesados de SPSS y caches del sistema
*.sav
*.dta
*.csv
*.parquet
.config/
sample_data/
__pycache__/
.ipynb_checkpoints/
"""
with open(".gitignore", "w", encoding="utf-8") as f:
    f.write(gitignore_content)
print("Archivo '.gitignore' configurado correctamente para la protección de datos.")
```

### Automatización del Despliegue de Entregables a GitHub

El siguiente script clona tu repositorio público de GitHub, copia de manera estructurada los entregables y archivos de control generados (`docs/`, `outputs/`, `requirements.txt`, `.gitignore`), y prepara el entorno para realizar el commit final de entrega.

```
import os
import shutil

REPO_URL = "https://github.com/limbertch26-alt/Repositorio-nuevo.git"
REPO_NAME = "Repositorio-nuevo"

# 1. Limpiar clones previos si existen
if os.path.exists(REPO_NAME):
    shutil.rmtree(REPO_NAME)
    print(f"Directorio previo {REPO_NAME} removido para evitar colisiones.")

# 2. Clonar el repositorio remoto
print(f"Clonando repositorio desde: {REPO_URL}...")
clone_status = os.system(f"git clone {REPO_URL}")

if clone_status == 0:
    print(f"¡Repositorio {REPO_NAME} clonado con éxito!")

    # 3. Estructurar carpetas internas en el repo clonado
    carpetas_repo = ["data", "notebooks", "src", "outputs", "docs"]
    for folder in carpetas_repo:
        os.makedirs(os.path.join(REPO_NAME, folder), exist_ok=True)

    # 4. Copiar archivos organizados al repositorio clonado
    copias = [
        ("requirements.txt", f"{REPO_NAME}/requirements.txt"),
        (".gitignore", f"{REPO_NAME}/.gitignore"),
        ("docs/reporte_ejecutivo_storytelling.html", f"{REPO_NAME}/docs/reporte_ejecutivo_storytelling.html"),
        ("docs/mapa_seguridad_alimentaria.html", f"{REPO_NAME}/docs/mapa_seguridad_alimentaria.html"),
        ("outputs/reporte_eda_interactivo.html", f"{REPO_NAME}/outputs/reporte_eda_interactivo.html"),
        ("outputs/clusters_pca_interactivo.html", f"{REPO_NAME}/outputs/clusters_pca_interactivo.html"),
        ("outputs/barras_agrupadas_seguridad.html", f"{REPO_NAME}/outputs/barras_agrupadas_seguridad.html"),
        ("outputs/sunburst_seguridad.html", f"{REPO_NAME}/outputs/sunburst_seguridad.html")
    ]

    for origen, destino in copias:
        if os.path.exists(origen):
            shutil.copy2(origen, destino)
            print(f"Copiado a repo: {origen} -> {destino}")

    print("\nTodo el espacio de trabajo ha sido estructurado y copiado al repositorio local.")
else:
    print("Error al clonar el repositorio. Verifica que la URL del repositorio remoto sea pública o tenga los permisos necesarios.")
```
